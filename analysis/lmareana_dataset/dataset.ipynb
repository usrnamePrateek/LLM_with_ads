# LMArena Human Preference: Dataset Exploration

Exploratory look at [`lmarena-ai/arena-human-preference-140k`](https://huggingface.co/datasets/lmarena-ai/arena-human-preference-140k),
the source data for the ad-placement pipeline.

Each row is a head-to-head "battle": one user prompt, two anonymous model responses
(`conversation_a` / `conversation_b`), and a human vote (`winner`).

**Goal:** reduce it to a clean set of **English, non-code, single-turn** `(query, llm_response)` pairs,
where `llm_response` is the response that won the vote.

| Step | Filter / transform |
|---|---|
| 1 | Drop coding conversations (`is_code == True`) |
| 2 | Keep English only (`language == "en"`) |
| 3 | Drop `both_bad` votes; resolve `tie` with a seeded coin flip |
| 4 | Keep the winning side's `model` + `conversation` |
| 5 | Keep single-turn conversations (1 user + 1 assistant message) |
| 6 | Flatten to plain-text `query` / `llm_response` |

> The production version of this logic lives in `src/ad_generation/arena.py`
> (`PrepareArenaDatasetService`, run via `src/ad_generation/scripts/prepare_dataset.py`).
> This notebook is the exploration that informed it.

## 1. Setup

`HF_TOKEN` is read from `.env` (see `.env.example`); the dataset is gated, so a token is required.

```python
import os
import random
from collections import Counter, defaultdict

from datasets import load_dataset
from dotenv import load_dotenv

load_dotenv()  # exposes HF_TOKEN via os.getenv
```

## 2. Load the dataset

First run downloads ~1.6 GB into the Hugging Face cache; later runs read from the cache.

```python
# Arena Human Preference 140k (~136k rows, ~1.6GB)
# https://huggingface.co/datasets/lmarena-ai/arena-human-preference-140k
ds = load_dataset(
    "lmarena-ai/arena-human-preference-140k",
    token=os.getenv("HF_TOKEN"),
)
```

### Schema

The dataset has a single `train` split. Key columns:

- `id`: unique battle id
- `model_a`, `model_b`: the two models compared
- `conversation_a`, `conversation_b`: message lists (`role` + list of typed `content` parts)
- `winner`: `model_a`, `model_b`, `tie`, or `both_bad`
- `language`, `is_code`: metadata used for filtering below

```python
train = ds["train"]
print(f"Rows: {len(train):,}")
train.features  # full column schema
```

```python
# Peek at one row (omit long conversation bodies for readability)
row = train[0]
{k: row[k] for k in ("id", "model_a", "model_b", "winner", "language", "is_code", "timestamp")}
```

## 3. Filter: non-code, English only

Coding prompts are a poor fit for ad placement, and the ad inventory and judge prompts are English,
so both are filtered out up front.

```python
# Step 1: keep only non-code conversations
train_no_code = train.filter(lambda x: not x["is_code"])
print(f"Rows after filtering out is_code=True: {len(train_no_code):,}")
```

```python
# Step 2: keep only English conversations
train_only_en = train_no_code.filter(lambda x: x["language"] == "en")
print(f"Rows after filtering out non-English: {len(train_only_en):,}")
```

## 4. Resolve the winner

We want one response per prompt: the one humans preferred. Check which vote labels exist,
and how many rows are ambiguous (`tie`) or unusable (`both_bad`).

```python
# Distinct vote labels
set(train_only_en["winner"])
```

```python
# Count the two ambiguous vote types
train_only_tie = train_only_en.filter(lambda x: x["winner"] == "tie")
train_only_both_bad = train_only_en.filter(lambda x: x["winner"] == "both_bad")

print(f"tie:      {len(train_only_tie):,}")
print(f"both_bad: {len(train_only_both_bad):,}")
```

**Decision:**

- `both_bad`: dropped, since neither response is a good stand-in for a real assistant answer.
- `tie`: kept; one side is picked at random. The RNG is seeded with the row `id`, so the pick is
  **deterministic** and reproducible across runs.

`add_winner_cols` collapses the A/B pair into a single `model` + `conversation`.

```python
# Step 3: drop both_bad; keep ties (pick a side at random below)
n_before = len(train_only_en)
train_no_both_bad = train_only_en.filter(lambda x: x["winner"] != "both_bad")
print(f"Dropped both_bad: {n_before - len(train_no_both_bad):,} | remaining: {len(train_no_both_bad):,}")


def add_winner_cols(row):
    """Step 4: return the winning side's model name and conversation."""
    winner = row["winner"]
    if winner == "model_a":
        side = "a"
    elif winner == "model_b":
        side = "b"
    else:  # tie — deterministic random choice from id
        side = random.Random(row["id"]).choice(["a", "b"])

    if side == "a":
        model = row["model_a"]
        conv = row["conversation_a"]
    else:
        model = row["model_b"]
        conv = row["conversation_b"]

    # Normalize schema (conversation_b has optional num_tokens)
    conversation = [{"role": m["role"], "content": m["content"]} for m in conv]
    return {"model": model, "conversation": conversation}


train_filtered = train_no_both_bad.map(add_winner_cols)
train_filtered.select_columns(["id", "winner", "model", "conversation"])[0]
```

## 5. Inspect conversation structure

Before flattening to plain text, check:

1. which `content` part types appear per role (text only, or also images, etc.)
2. how many messages a conversation has (single-turn vs multi-turn)

```python
# Unique content part types per role in the winning conversation
types_by_role = defaultdict(set)

for conv in train_filtered["conversation"]:
    for msg in conv:
        role = msg["role"]
        for part in msg["content"]:
            types_by_role[role].add(part["type"])

{role: sorted(types) for role, types in types_by_role.items()}
```

```python
# Messages per conversation (2 = one user turn + one assistant turn)
lengths = [len(conv) for conv in train_filtered["conversation"]]
print(f"max blocks: {max(lengths)}")
print(f"min blocks: {min(lengths)}")
print(f"mean blocks: {sum(lengths) / len(lengths):.2f}")
```

```python
# Distribution of conversation lengths: {num_messages: num_rows}
counts = Counter(lengths)
counts
```

Multi-turn rows alternate `user` / `assistant`. Look at a 4-message example to see what a
follow-up turn looks like (first 300 chars of each message).

```python
# Example row with exactly 4 conversation blocks
idx = next(i for i, n in enumerate(lengths) if n == 4)
row = train_filtered[idx]

print({k: row[k] for k in ("id", "winner", "model", "language")})
print(f"blocks: {len(row['conversation'])}")
for i, msg in enumerate(row["conversation"]):
    text = msg["content"][0]["text"] if msg["content"] else ""
    preview = text[:300] + ("..." if len(text) > 300 else "")
    print(f"\n[{i}] role={msg['role']}\n{preview}")
```

## 6. Keep single-turn conversations

Follow-up turns depend on earlier context, which makes ad matching on "the query" ambiguous.
Keeping only 2-message rows gives a clean one-query → one-response pair.

```python
# Step 5: exactly one user message + one assistant message
train_small_conv = train_filtered.filter(lambda x: len(x["conversation"]) == 2)
len(train_small_conv)
```

## 7. Flatten to `query` / `llm_response`

Concatenate the `text` parts of each message (non-text parts such as images are skipped).
This produces the `(query, llm_response)` columns used downstream.

```python
def _text_from_msg(msg):
    """Join all text parts of a message; `text` can be None, so coerce to ""."""
    return "".join(part["text"] or "" for part in msg["content"] if part["type"] == "text")


def add_query_response(row):
    """Step 6: extract the user query and the winning assistant response as plain text."""
    query = ""
    llm_response = ""
    for msg in row["conversation"]:
        if msg["role"] == "user":
            query = _text_from_msg(msg)
        elif msg["role"] == "assistant":
            llm_response = _text_from_msg(msg)
    return {"query": query, "llm_response": llm_response}


train_small_conv = train_small_conv.map(add_query_response)
train_small_conv.select_columns(["id", "model", "query", "llm_response"])[0]
```

## 8. Model coverage

Which model families produced the winning responses in the final set? Model names are grouped
by the prefix before the first `-` (e.g. `gpt-4o-...` → `gpt`, `claude-3-5-...` → `claude`).

```python
# Rows per model family, most frequent first
prefixes = [m.split("-", 1)[0] for m in train_small_conv["model"]]
rows_per_model = Counter(prefixes)
sorted(rows_per_model.items(), key=lambda x: (-x[1], x[0]))
```
